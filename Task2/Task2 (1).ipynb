```python
region_sales = df.groupby("Region")["Sales"].sum()
region_sales
```

```python
region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()
category_sales
```

```python
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
df.columns
```

```python
subcategory_sales = df.groupby("Sub_Category")["Sales"].sum()
subcategory_sales.sort_values(ascending=False)
```

```python
subcategory_sales.sort_values().plot(kind="barh")
plt.title("Sales by Sub-Category")
plt.xlabel("Sales")
plt.ylabel("Sub-Category")
plt.show()
```

```python
top_products = df.groupby("Product_Name")["Sales"].sum().sort_values(ascending=False).head(10)
top_products
```

```python
top_products.plot(kind="bar")
plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
low_products = df.groupby("Product_Name")["Sales"].sum().sort_values().head(10)
low_products
```

```python
import seaborn as sns

```

```python
df.columns = df.columns.str.strip()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="%d/%m/%Y")
```

```python
df["Order_Date"].head()
```

```python
monthly_sales = df.groupby(df["Order_Date"].dt.month)["Sales"].sum()
monthly_sales
```

```python
monthly_sales.plot(kind="line", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
sns.histplot(df["Sales"], bins=30)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
df.corr(numeric_only=True)
```

```python
total_revenue = df["Sales"].sum()
total_revenue
```

```python
total_orders = df["Order_ID"].nunique()
total_orders
```

```python
average_order_value = total_revenue / total_orders
average_order_value
```

```python
print("Total Revenue:", total_revenue)
print("Total Orders:", total_orders)
print("Average Order Value:", average_order_value)
```

# Recommendations
