```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df["Product Category"].value_counts()
```

```python
df['Product Category'].value_counts().plot(kind='bar')
plt.title('Product Category Distribution')
plt.xlabel('Product Category')
plt.ylabel('Count')
plt.show()
```

```python
df.columns
```

```python
df["Gender"].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title('Gender distribution')
plt.ylabel(" ")
plt.show()
```

```python
df['Customer Age'].plot(kind='hist', bins=10)
plt.title('Customer Age Distribution')
plt.xlabel('Customer Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Customer Age'], df['Total Purchase Amount'])
plt.title('Customer Age vs Total Purchase Amount')
plt.xlabel('Customer Age')
plt.ylabel('Total Purchase Amount')
plt.show()
```

```python
df.groupby("Gender")["Total Purchase Amount"].mean()
```

```python
df[['Customer ID', 'Total Purchase Amount']].sort_values(by='Total Purchase Amount', ascending=False).head(10)
```

```python
df.groupby('Product Category')['Total Purchase Amount'].mean().sort_values(ascending=False)
```

```python
df['Payment Method'].value_counts().plot(kind='bar')
plt.title('Payment Method Distribution')
plt.xlabel('Payment Method')
plt.ylabel('Number of Customers')
plt.show()
```

```python
df['Churn'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Customer Churn Distribution')
plt.ylabel('')
plt.show()
```
